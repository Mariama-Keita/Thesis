```python
import importlib
import parsing
importlib.reload(parsing)
from parsing import parse_medium
import pandas as pd
```

```python
clean = pd.read_csv("/Users/admin/desktop/medium_clean.csv")
```

```python
all_ins, all_comp = [], []

for _, row in clean.iterrows():
    ins, comp = parse_medium(row["grmd"], row["tex_text_clean"])
    all_ins.extend(ins)
    all_comp.extend(comp)

instr_df = pd.DataFrame(all_ins)
comp_df  = pd.DataFrame(all_comp)
# ensure `amount` is numeric for aggregation/sorting; invalid values become NaN
comp_df["amount"] = pd.to_numeric(comp_df["amount"], errors="coerce")

# turn (grmd, step) into a readable step number
comp_df["step"] = comp_df["instruction_id"].apply(lambda x: x[1] if x else None)
comp_df = comp_df.drop(columns=["instruction_id"])

print("instructions:", len(instr_df))
print("compositions:", len(comp_df))
print()
print(instr_df["kind"].value_counts())
print()
print("base components (no step):", comp_df["step"].isna().sum())
```
